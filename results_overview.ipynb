```python


import pandas as pd
import os

#Define the file path
file_path = "ragas_evaluation_results.csv"
if os.path.exists(file_path):
    results = pd.read_csv(file_path)
    
    # Ensure these columns actually exist in the CSV to avoid crashes
    metrics_cols = ['faithfulness', 'answer_relevancy', 'context_precision', 'context_recall']
    available_cols = [c for c in metrics_cols if c in results.columns]
    
    if 'config_id' in results.columns and available_cols:
        summary_df = results.groupby('config_id')[available_cols].mean()
        print("\n--- Average Scores by Configuration ---")
        print(summary_df)
    else:
        print("Error: Required columns (config_id or metrics) not found in CSV.")
else:
    print(f"Error: The file '{file_path}' was not found in the current directory.")
```
